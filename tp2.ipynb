IMPORTO LA LIBRERIA PANDAS

```python
import pandas as pd
```

LEO EL ARCHIVO Y LO TRANFORMO EN UN DATAFRAME 

```python
transacciones = pd.read_csv(r"C:\Users\sebas\Desktop\data science 2\TP2\transacciones.csv")

print(transacciones)
```

AHORA CARGO EL ARCHIVO EXCEL 

```python
productos = pd.read_excel(
    r"C:\Users\sebas\Desktop\data science 2\TP2\productos.xlsx",
    engine="openpyxl"
)

print(productos)
```

compruebo que los datos sean correctos

```python
print(transacciones.dtypes)
print(productos.dtypes)
```

convierto fecha a datetime

```python
transacciones["fecha"] = pd.to_datetime(
    transacciones["fecha"]
)
```

verifico

```python
print(transacciones.dtypes)
```

aseguro que los ID sean compatibles

```python
transacciones["id_producto"] = transacciones["id_producto"].astype(int)

productos["id_producto"] = productos["id_producto"].astype(int)
```

busco valores nulos 

```python
print(transacciones.isnull().sum())
print(productos.isnull().sum())
```

hay 1 valor nulo en cantidad , en nuestro caso tiene ssentido eliminar esa transaccion porque no sabemos cuantas unidades fueron vendidas

```python
transacciones = transacciones.dropna(
    subset=["cantidad"]
)
```

convierto cantidad de float a entero 

```python
transacciones["cantidad"] = transacciones["cantidad"].astype(int)
```

ahora hacemos el merge 

```python
ventas = pd.merge(
    transacciones,
    productos,
    on="id_producto",
    how="left"
)
```

comprobamos

```python
print(ventas)
```

calcular total_venta

```python
ventas["total_venta"] = (
    ventas["cantidad"] * ventas["precio_unitario"]
)
```

verifico

```python
print(ventas)
```

Exportar a Parquet

```python
ventas.to_parquet(
    "ventas_normalizadas.parquet",
    index=False
)
```
